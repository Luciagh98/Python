### 2.7. Función `procesar_datos`.

### Explicación del ejercicio:

Crea una función llamada `procesar_datos` que procesa una serie de datos dependiendo del tipo de operación especificada.

### Pasos a seguir:

1. **Definición de la función `procesar_datos`:**
   - Creamos una función llamada `procesar_datos` que acepta una operación como primer argumento (`sumar`, `restar`, `multiplicar`, `dividir`) y un número arbitrario de argumentos posicionales (`*args`) que representan los datos sobre los cuales se realizará la operación. Además, se pueden proporcionar argumentos de palabra clave (`**kwargs`) que especifican opciones adicionales para la operación, como el valor inicial para la suma o multiplicación.

2. **Implementación de la lógica de las operaciones:**
   - Utilizamos una estructura condicional para determinar la operación a realizar según el valor de `operacion`.
     - Si `operacion` es `'sumar'`, sumamos los datos proporcionados, considerando un posible valor inicial para la suma.
     - Si `operacion` es `'restar'`, restamos los datos proporcionados.
     - Si `operacion` es `'multiplicar'`, multiplicamos los datos proporcionados, considerando un posible factor de multiplicación.
     - Si `operacion` es `'dividir'`, dividimos los datos proporcionados.
     - En caso de que `operacion` no sea ninguna de las operaciones válidas, se levanta una excepción.

3. **Retorno del resultado:**
   - Devolvemos el resultado de la operación realizada sobre los datos.

### Ejemplo de uso:


- resultado_suma = procesar_datos('sumar', 1, 2, 3, 4, valor_inicial=10)


- resultado_resta = procesar_datos('restar', 20, 5, 3)


- resultado_multiplicacion = procesar_datos('multiplicar', 2, 3, 4, factor=2)


- resultado_division = procesar_datos('dividir', 100, 5, 2)

### 2.11. Mapear una lista.

### Descripción del ejercicio:

Crea una lista de números y utiliza una lambda junto con la función `map()` para elevar al cuadrado cada número en la lista.

### Pasos a seguir:

1. **Definición de la lista de números:**
   - Creamos una lista de números llamada `numbers` que contiene los números a mapear.
   - Ejemplo de uso: 
    - numbers = [1, 2, 3, 4, 5]
  

2. **Mapeo de la lista:**
   - Utilizamos la función `map()` junto con una lambda para aplicar una función a cada elemento de la lista.
   - La lambda `lambda x: x**2` eleva al cuadrado cada número `x`.
   - Convertimos el resultado de `map()` en una lista para obtener los números elevados al cuadrado.
   - Ejemplo de uso:
     ```python
     squared_numbers = list(map(lambda x: x**2, numbers))
     ```

3. **Imprimimos la lista resultante:**
   - Imprimimos la lista resultante que contiene los números elevados al cuadrado.

   - Convertimos el resultado de `filter()` en una lista usando el método `list()` para obtener los números que cumplen la condición.
   

3. **Contamos los elementos filtrados:**
   - Utilizamos la función `len()` para contar cuántos elementos hay en la lista filtrada

   - Imprimimos la cantidad de números que son mayores que 5.
