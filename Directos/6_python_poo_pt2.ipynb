### 2.7. Métodos de interacción del trabajador con el sistema de comercio electrónico
###  Explicación del ejercicio:

En este ejercicio, añadiremos métodos a la clase `ECommerce` para permitir la interacción del trabajador con el sistema de comercio electrónico, como agregar productos al inventario, ver información detallada de un producto y ver el inventario completo.

### Código a seguir:

1. **Método para agregar productos al inventario:**
   - Creamos un método llamado `agregar_al_inventario` que toma un objeto `producto` como argumento y lo agrega al inventario utilizando el título en minúsculas como clave.

2. **Método para ver información detallada de un producto en el inventario:**
   - Creamos un método llamado `ver_producto` que toma el `nombre_producto` como argumento.
   - Eliminamos los espacios en blanco al principio y al final del nombre del producto y lo convertimos a minúsculas.
   - Verificamos si el producto está disponible en el inventario.
   - Si el producto está en el inventario, mostramos su información detallada.
   - Si no, mostramos un mensaje de error.

3. **Método para ver el inventario completo:**
   - Creamos un método llamado `ver_inventario` que muestra el inventario completo.
   - Recorremos todos los productos en el inventario y mostramos la información detallada de cada uno.

### 2.8. Interacción del cliente con el sistema de comercio electrónico
###  Explicación del ejercicio:

En este ejercicio, añadiremos métodos a la clase `ECommerce` para permitir que el cliente interactúe con el sistema de comercio electrónico, como agregar productos al carrito de compras, calcular el precio total de la compra, procesar el pago y ver los productos en el carrito.

### Código a seguir:

1. **Método para agregar productos al carrito de compras:**
   - Creamos un método llamado `agregar_al_carrito` que toma el `nombre_producto` como argumento.
   - Eliminamos los espacios en blanco al principio y al final del nombre del producto y lo convertimos a minúsculas.
   - Verificamos si el producto está en el inventario y si hay suficientes unidades disponibles.
   - Si el producto está disponible, lo agregamos al carrito de compras, reducimos la cantidad disponible en el inventario y mostramos un mensaje de éxito.
   - Si no hay suficientes unidades disponibles, mostramos un mensaje de error.
   - Si el producto no está en el inventario, mostramos un mensaje de error.

2. **Método para calcular el precio total de los productos en el carrito:**
   - Creamos un método llamado `calcular_precio_total` que calcula la suma de los precios de todos los productos en el carrito.

3. **Método para procesar el pago de los productos en el carrito:**
   - Creamos un método llamado `pagar` que toma el `monto` como argumento.
   - Verificamos si el monto ingresado es suficiente para pagar el total de la compra.
   - Si es suficiente, calculamos el cambio, mostramos un mensaje de éxito, vaciamos el carrito y mostramos el cambio.
   - Si el monto es insuficiente, mostramos un mensaje de error.
   - Si se ingresa un monto no válido, mostramos un mensaje de error.

4. **Método para ver los productos en el carrito de compras:**
   - Creamos un método llamado `ver_carrito` que muestra los productos en el carrito de compras.
   - Si hay productos en el carrito, mostramos los títulos de los productos y el precio total a pagar.
   - Si el carrito está vacío, mostramos un mensaje indicando que está vacío.

### 2.9. Poniendo en uso todo lo anterior.
###  Explicación del ejercicio:
En este último ejercicio vamos a probar las funcionalidades de nuestro e-commerce. Para ello tendremos que simular primero que somos un trabajador y luego un cliente que compra nuestros libros.

### Código a seguir:

1. **Creación de instancia de ECommerce:**
   - Creamos una instancia de la clase `ECommerce` llamada `ecommerce`.

2. **Ingreso de libros al inventario (simulado como trabajador):**
   - Creamos instancias de libros físicos y audiolibros.
   - Solicitamos al usuario ingresar la cantidad de libros disponibles (esto podría ser reemplazado con una funcionalidad más automatizada).
   - Agregamos los libros al inventario utilizando el método `agregar_al_inventario` de la clase `ECommerce`.

3. **Visualización del inventario:**
   - Mostramos el inventario actual utilizando el método `ver_inventario` de la clase `ECommerce`.

4. **Simulación de la compra por parte de un cliente:**
   - Creamos un bucle que solicita al cliente que elija un producto para agregar al carrito o finalice la compra.
   - Si el cliente elige finalizar la compra, mostramos los productos en el carrito, solicitamos al cliente ingresar el monto para realizar el pago y procesamos el pago utilizando el método `pagar` de la clase `ECommerce`.
   - Si el cliente elige agregar un producto, lo agregamos al carrito de compras utilizando el método `agregar_al_carrito` de la clase `ECommerce`.

5. **Muestra de productos en el carrito y actualización del inventario:**
   - Después de realizar la compra, mostramos nuevamente los productos en el carrito de compras utilizando el método `ver_carrito` de la clase `ECommerce`.
   - Mostramos el inventario completo después de la compra utilizando el método `ver_inventario` de la clase `ECommerce`.
