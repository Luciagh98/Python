## 1. Resumen del reto guiado de Python: Programación Orientada a Objetos (Parte 1)


### 1.1. Conocimientos previos.
- Python: Variables, tipos de datos y estructuras de datos, condidicionales, bucles `for` y `while`.
- Python: Programación Funcional.
- Python: Programación Orientada a Objetos. Clases, método constructor, métodos propios, Herencia, Polimorfismo.

### 1.2. Propósito del reto.
Nuestra empresa nos ha pedido que diseñemos un script de python para establecer el funcionamiento de un ECommerce para la venta de libros. Para ello seguiremos varios pasos
- **Primero tendremos que determinar el comportamiento de diferentes clases de libros.**
- Segundo generaremos un ECommerce que permita a clientes y trabajadores interaccionar con los libros de los que disponemos.

### 2.2. Métodos adicionales para la clase **Libro**.
###  Explicación del ejercicio:

En este ejercicio, ampliaremos la clase `Libro` con métodos adicionales para establecer la cantidad de libros disponibles y obtener información detallada del libro.

 ### Código a seguir:
1. **Definición del método para establecer la cantidad de libros disponibles:**
   - Definimos un método llamado `establecer_cantidad()` dentro de la clase `Libro` para permitir al usuario establecer la cantidad de libros disponibles.
   - Este método solicitará al usuario que ingrese la cantidad de libros y la asignará al atributo `cantidad` del libro.

2. **Definición del método para obtener información detallada del libro:**
   - Definimos un método llamado `info_libro()` dentro de la clase `Libro` para obtener información detallada del libro.
   - Este método devolverá una cadena con el título, autor, precio y cantidad disponible del libro.

3. **Instanciación y llamada a métodos:**
   - Creamos una nueva instancia de la clase Libro.
   - Llamamos al método `establecer_cantidad` para establecer la cantidad de libros disponibles.
   - Llamamos al método `info_libro` para obtener información sobre el libro.

### 2.4. Métodos adicionales para las clases **LibroFisico** y **AudioLibro**.
###  Explicación del ejercicio:

En este ejercicio, agregaremos métodos adicionales a las clases `LibroFisico` y `AudioLibro` para mejorar su funcionalidad.

 ### Código a seguir:
1. **Definición del método para calcular la fecha estimada de entrega en la clase LibroFisico:**
   - Agregamos un método llamado `envio()` dentro de la clase `LibroFisico` para calcular la fecha estimada de entrega.
   - Este método toma un parámetro opcional `dias_envio` que representa el número de días estimados para el envío. Pondremos 5 por defecto.
   - Utiliza las clases `datetime` y `timedelta` del módulo `datetime` para realizar cálculos de fechas.
   - Retorna la fecha estimada de entrega.

2. **Definición del método para convertir la duración del audiolibro en formato HH:MM:SS en la clase AudioLibro:**
   - Agregamos un método llamado `obtener_duracion()` dentro de la clase `AudioLibro` para convertir la duración del audiolibro en formato HH:MM:SS.
   - Este método calcula las horas, minutos y segundos a partir de la duración total en segundos y los concatena en un formato de hora.
   - Retorna la duración en formato HH:MM:SS.

### 2.5. Polimorfismo en las clases **LibroFisico** y **AudioLibro**.
###  Explicación del ejercicio:

En este ejercicio, implementaremos polimorfismo en las clases `LibroFisico` y `AudioLibro` al sobrescribir el método `info_libro()` para proporcionar información detallada específica de cada tipo de libro.

 ### Código a seguir:
1. **Definición del método `info_libro()` en la clase LibroFisico:**
   - Sobrescribimos el método `info_libro()` en la clase `LibroFisico` para incluir información adicional como la categoría del libro físico, el peso y la fecha estimada de entrega.
   - Llamamos al método `info_libro()` de la clase padre para obtener la información básica del libro.
   - Agregamos la categoría del libro físico, el peso y la fecha de entrega estimada a la cadena de información.
   - Retornamos la cadena con la información del libro físico.

2. **Definición del método `info_libro()` en la clase AudioLibro:**
   - Sobrescribimos el método `info_libro()` en la clase `AudioLibro` para incluir información adicional como la categoría del audiolibro, la duración en segundos y la duración en formato HH:MM:SS.
   - Llamamos al método `info_libro()` de la clase padre para obtener la información básica del libro.
   - Agregamos la categoría del audiolibro, la duración en segundos y la duración en formato HH:MM:SS a la cadena de información.
   - Retornamos la cadena con la información del audiolibro.
